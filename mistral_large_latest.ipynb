### Mistral large latest model test

```python
import os
from mistralai import Mistral, UserMessage #biblio Mistral 
from dotenv import load_dotenv #lecture des keys du modèle
```

```python
load_dotenv()
```

```python
api_key = os.environ.get("API_KEY") #ici - l'API Mistral conservée dans le fichier .env dans une variable API_KEY
if not api_key : 
    raise ValueError("API_KEY is not set")
```

```python
model_name = "mistral-large-latest" #ici on choisit le modèle qui sera utilisé plus bas 
```

```python
client = Mistral(api_key = api_key) #ici le bot 
print(f"Send a request to the model : {model_name}") #check si le request a été envoyé 
```

```python
api_key = os.environ["API_KEY"]
model = model_name
client = client

#la boucle pour la conversation continue 
while True: 
    user_input = input("mon message :") #la fonction pour les messages entrants

    chat_response = client.chat.complete(
    model = model_name, 
    messages = [
        {
            "role" : "user", 
            "content" : user_input,
        }
    ]
)

    if user_input == ('stop'): #ici on arrête la conversation 
        break

    print(chat_response.choices[0].message.content)
```
