```python
import os
from mistralai import Mistral
from dotenv import load_dotenv 
from langchain_mistralai import ChatMistralAI
from langchain_mistralai.chat_models import ChatMistralAI
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
load_dotenv() #ici on cherhche la clé API dans .env
```

```python
api_key = os.environ.get("MISTRAL_API_KEY") #ici on reprend la clé Mistral depuis les variables d'environement
if not api_key : 
    raise ValueError("MISTRAL_API_KEY is not set")

llm = ChatMistralAI( #initialisation du modèle de chat Mistral 
    model = "mistral-large-latest", 
    temperature = 0.7, #ici on choisit la variativité des réponses (voir ci-dessous)
    max_retries = 2, #nombre des tentantives en cas d'échec de requête
    mistral_api_key = api_key #vérif que la clé API soit bien transmise
)
```

```python
memory = ConversationBufferMemory() #création de la mémo pour stocker la conversation 
```

```python
conversation = ConversationChain( #ici on crée la chîne de conv une fois dehors de la boucle pour ne pas bouffer la mémoire 
        llm = llm, 
        memory = memory
    )
```

```python
while True: 
    user_input = input("mon message :") #ici on démarre la conversation 
    if user_input == ('stop'): #pour finir la conversation et que la cullule ne tourne pas 
        break
   
    response = conversation.run(user_input) #ici on envoie le msg à la chaîne de conv et récupère la réponse
    print(response)
```
